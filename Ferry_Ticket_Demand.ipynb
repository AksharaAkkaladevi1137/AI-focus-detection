````python
df_model = df_model.set_index('Timestamp')

sales = df_model['Sales Count']
redemption = df_model['Redemption Count']

# Lag features
df_model['Sales_Lag_15m'] = sales.reindex(
    df_model.index - pd.Timedelta(minutes=15)
).values

df_model['Sales_Lag_30m'] = sales.reindex(
    df_model.index - pd.Timedelta(minutes=30)
).values

df_model['Sales_Lag_60m'] = sales.reindex(
    df_model.index - pd.Timedelta(minutes=60)
).values

df_model['Sales_Lag_120m'] = sales.reindex(
    df_model.index - pd.Timedelta(minutes=120)
).values


df_model['Redemption_Lag_15m'] = redemption.reindex(
    df_model.index - pd.Timedelta(minutes=15)
).values

df_model['Redemption_Lag_30m'] = redemption.reindex(
    df_model.index - pd.Timedelta(minutes=30)
).values

df_model['Redemption_Lag_60m'] = redemption.reindex(
    df_model.index - pd.Timedelta(minutes=60)
).values

df_model['Redemption_Lag_120m'] = redemption.reindex(
    df_model.index - pd.Timedelta(minutes=120)
).values
````

````python
df_model['Sales_Target_15m'] = sales.reindex(
    df_model.index + pd.Timedelta(minutes=15)
).values

df_model['Redemption_Target_15m'] = redemption.reindex(
    df_model.index + pd.Timedelta(minutes=15)
).values
````

````python
model_15m = df_model.copy()

features_15m = [
    'Sales_Lag_15m',
    'Sales_Lag_30m',
    'Sales_Lag_60m',
    'Sales_Lag_120m',
    'Redemption_Lag_15m',
    'Redemption_Lag_30m',
    'Redemption_Lag_60m',
    'Redemption_Lag_120m',
    'Hour',
    'DayOfWeek',
    'Month',
    'Day',
    'IsWeekend'
]

target_15m = 'Sales_Target_15m'

model_15m = model_15m[
    ['Timestamp'] + features_15m + [target_15m]
].dropna()

print("Original rows:", len(df_model))
print("15-minute modeling rows:", len(model_15m))
print("Rows removed:", len(df_model) - len(model_15m))
print("Remaining %:", round(len(model_15m) / len(df_model) * 100, 2))

print(model_15m.isna().sum())
````

````python
split_index = int(len(model_15m) * 0.8)

train = model_15m.iloc[:split_index].copy()
test = model_15m.iloc[split_index:].copy()

print("Train shape:", train.shape)
print("Test shape:", test.shape)

print("\nTraining period:")
print(train['Timestamp'].min())
print("to")
print(train['Timestamp'].max())

print("\nTesting period:")
print(test['Timestamp'].min())
print("to")
print(test['Timestamp'].max())
````

````python
features_15m = [
    'Sales_Lag_15m',
    'Sales_Lag_30m',
    'Sales_Lag_60m',
    'Sales_Lag_120m',
    'Redemption_Lag_15m',
    'Redemption_Lag_30m',
    'Redemption_Lag_60m',
    'Redemption_Lag_120m',
    'Hour',
    'DayOfWeek',
    'Month',
    'Day',
    'IsWeekend'
]

X_train = train[features_15m]
y_train = train['Sales_Target_15m']

X_test = test[features_15m]
y_test = test['Sales_Target_15m']

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
````

````python
def forecast_next_15_minutes(df, rolling_features, model, feature_columns):
    latest_time = df["Timestamp"].max()
    forecast_time = latest_time + pd.Timedelta(minutes=15)

    row = {
        "Timestamp": forecast_time,
        "Hour": forecast_time.hour,
        "DayOfWeek": forecast_time.dayofweek,
        "Month": forecast_time.month,
        "Day": forecast_time.day,
        "IsWeekend": int(forecast_time.dayofweek >= 5)
    }

    lag_minutes = [15, 30, 60, 120]

    for minutes in lag_minutes:
        lag_time = latest_time - pd.Timedelta(minutes=minutes)
        lag_data = df[df["Timestamp"] == lag_time]

        if len(lag_data) > 0:
            row[f"Sales_Lag_{minutes}m"] = lag_data["Sales Count"].iloc[0]
            row[f"Redemption_Lag_{minutes}m"] = lag_data["Redemption Count"].iloc[0]
        else:
            row[f"Sales_Lag_{minutes}m"] = np.nan
            row[f"Redemption_Lag_{minutes}m"] = np.nan

    latest_rolling = rolling_features[
        rolling_features["Timestamp"] == latest_time
    ]

    if len(latest_rolling) > 0:
        for column in latest_rolling.columns:
            if column != "Timestamp":
                row[column] = latest_rolling[column].iloc[0]

    forecast_input = pd.DataFrame([row])

    for column in feature_columns:
        if column not in forecast_input.columns:
            forecast_input[column] = np.nan

    forecast_input = forecast_input[feature_columns]

    prediction = model.predict(forecast_input)[0]

    return {
        "Latest observed time": latest_time,
        "Forecast time": forecast_time,
        "Predicted Sales Count": round(float(prediction), 2)
    }


forecast_result = forecast_next_15_minutes(
    df_clean,
    rolling_features,
    model_xgb_v2,
    features_15m_v2
)

forecast_result
````

````python
%%writefile README.md
# Ferry Ticket Demand Forecasting

## Project Overview

This project analyzes historical ferry ticket sales and redemption data and forecasts ferry ticket demand for the next 15 minutes.

The project uses historical demand patterns, time-based features, lag features, and rolling demand features to build a short-term forecasting model.

A Streamlit dashboard is included to display the latest demand, forecast results, trends, anomalies, and model performance.

---

## Dataset

The dataset contains ferry ticketing records with the following main columns:

- `_id`
- `Timestamp`
- `Sales Count`
- `Redemption Count`

The dataset contains approximately 261,538 records covering data from 2015 to 2025.

---

## Data Preprocessing

The following preprocessing steps were performed:

- Converted `Timestamp` into datetime format.
- Extracted time-based features:
  - Year
  - Month
  - Day
  - Day of Week
  - Hour
  - Is Weekend
- Checked for missing time intervals.
- Checked for negative sales and redemption values.
- Investigated unusually large demand values.

---

## Anomaly Detection

Unusually large observations were identified using a threshold of greater than 1000 sales or redemptions.

These observations were investigated rather than automatically removed because they may represent genuine operational events or data aggregation issues.

The anomalies were also evaluated separately during model validation.

---

## Feature Engineering

The forecasting model uses historical demand information through lag features.

Lag features include:

- Sales at 15 minutes earlier
- Sales at 30 minutes earlier
- Sales at 60 minutes earlier
- Sales at 120 minutes earlier
- Redemption values at the same intervals

Rolling demand features were also created:

- 1-hour rolling sales mean
- 1-hour rolling sales maximum
- 2-hour rolling sales mean
- 2-hour rolling sales maximum
- 1-hour rolling redemption mean
- 2-hour rolling redemption mean

These features help the model capture recent demand trends.

---

## Models Compared

The following forecasting approaches were evaluated:

1. Naive Forecast
2. Random Forest
3. XGBoost
4. XGBoost with Rolling Features

The final model selected for the dashboard is:

**XGBoost with Rolling Features**

---

## Model Performance

| Model | MAE | RMSE |
|---|---:|---:|
| Naive Forecast | 28.79 | 103.49 |
| Random Forest | 26.42 | 89.26 |
| XGBoost | 25.40 | 86.86 |
| XGBoost + Rolling Features | 21.55 | 74.33 |

The final model achieved:

- **MAE:** 21.55
- **RMSE:** 74.33

For normal-demand observations where the target was 1000 or below:

- **MAE:** 20.55
- **RMSE:** 36.82

---

## 15-Minute Forecasting

The trained model is used to predict ferry sales for the next 15-minute interval.

The forecasting pipeline uses the latest available observation together with the required lag, rolling, and calendar features.

A reusable forecasting function was created so that the model can generate a new prediction from the latest available data.

---

## Dashboard

A Streamlit dashboard was created to provide a visual interface for the forecasting system.

The dashboard displays:

- Latest Sales Count
- Latest Redemption Count
- Next 15-Minute Sales Forecast
- Number of anomalous records
- Forecast details
- Recent demand trends
- Average sales by hour
- Model performance
- Model feature information

---

## Project Structure

```text
Ferry Ticket Demand Forecasting/
│
├── forecasting_artifacts/
│   ├── xgboost_rolling_model.pkl
│   ├── feature_columns.pkl
│   └── cleaned_ferry_data.csv
│
├── dashboard.py
├── README.md
└── dataset
````
